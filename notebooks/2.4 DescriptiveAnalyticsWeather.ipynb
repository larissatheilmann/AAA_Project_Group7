# Descriptive Analytics - Weather Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# weather data
weather_df = pd.read_parquet("../data/weather_data.parquet")
weather_df.head()
```

```python
features = [
    "2m_temp",
    "total_precip",
    "snow_cov",
    "snow_depth",
    "wind_speed"
]

feature_labels = {
    "2m_temp": "2m Temperature",
    "total_precip": "Total Precipitation",
    "snow_cov": "Snow Coverage (%)",
    "snow_depth": "Snow Depth"
}

# Mapping of feature and color
colors = {
    "2m_temp": "red",
    "total_precip": "blue",
    "snow_cov": "cyan",
    "snow_depth": "purple"
}
```

```python
n_features = len(features)
fig, axes = plt.subplots(n_features, 1, figsize=(16, 3 * n_features), sharex=True)

for ax, feature in zip(axes, features):
    ax.plot(
        weather_df["time_step"],
        weather_df[feature],
        color=colors[feature],
        label=feature_labels[feature],
        linewidth=1.2
    )
    ax.set_title(feature_labels[feature], fontsize=12)
    ax.set_ylabel("Value")
    ax.legend(loc="upper right")
    ax.grid(True, alpha=0.3)
axes[-1].set_xlabel("Time")
fig.suptitle("Time Series of each Feature", fontsize=18, y=0.995)
plt.tight_layout()
plt.show()
```
